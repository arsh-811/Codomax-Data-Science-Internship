```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

print("All libraries imported successfully!")
```

```python
import pandas as pd

data = {
    'Student': ['Arun', 'Kumar', 'Priya', 'Rahul',
                'Divya', 'Manoj', 'Anitha', 'Sanjay'],
    'Marks': [85, 90, 78, 92, 88, 65, 95, 72],
    'Attendance': [90, 95, 85, 98, 88, 70, 96, 80],
    'Department': ['CSE', 'IT', 'AI&DS', 'CSE',
                   'IT', 'AI&DS', 'CSE', 'IT']
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(9, 5))

plt.bar(df['Student'], df['Marks'])

plt.title("Student Marks Comparison")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(df['Student'], df['Marks'], marker='o')

plt.title("Student Marks Trend")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
department_count = df['Department'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    department_count,
    labels=department_count.index,
    autopct='%1.1f%%'
)

plt.title("Student Department Distribution")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(df['Marks'], bins=5, edgecolor='black')

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Attendance',
    y='Marks',
    hue='Department',
    s=100
)

plt.title("Attendance vs Student Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")

plt.show()
```
