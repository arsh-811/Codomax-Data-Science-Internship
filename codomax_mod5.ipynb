```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Titanic Data Analysis Project")
print("All libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("First 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.round(2))
```

```python
# Fill missing age values with median
df['age'] = df['age'].fillna(df['age'].median())

# Fill missing embarked values with mode
df['embarked'] = df['embarked'].fillna(
    df['embarked'].mode()[0]
)

# Fill missing embark_town values with mode
df['embark_town'] = df['embark_town'].fillna(
    df['embark_town'].mode()[0]
)

print("Data cleaning completed!")

print("\nRemaining Missing Values:")
print(df.isnull().sum())
```

```python
survival_count = df['survived'].value_counts()

print("Survival Count:")
print(survival_count)

survival_rate = df['survived'].mean() * 100

print("\nOverall Survival Rate:",
      round(survival_rate, 2), "%")
```

```python
gender_survival = df.groupby('sex')['survived'].mean() * 100

print("Gender-wise Survival Rate:")
print(gender_survival.round(2))
```

```python
class_survival = df.groupby('pclass')['survived'].mean() * 100

print("Passenger Class Survival Rate:")
print(class_survival.round(2))
```

```python
print("Average Passenger Age:", round(df['age'].mean(), 2))

print("Youngest Passenger:", df['age'].min())

print("Oldest Passenger:", df['age'].max())
```

```python
print("Average Ticket Fare:", round(df['fare'].mean(), 2))

print("Highest Ticket Fare:", df['fare'].max())

print("Lowest Ticket Fare:", df['fare'].min())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='survived')

plt.title("Titanic Passenger Survival Count")
plt.xlabel("Survival (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='sex', hue='survived')

plt.title("Survival Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Passenger Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='pclass', y='survived')

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='age', bins=20, kde=True)

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Passenger Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='fare')

plt.title("Passenger Fare Distribution")
plt.xlabel("Ticket Fare")

plt.show()
```

```python
print("=" * 45)
print("TITANIC DATA ANALYSIS - KEY INSIGHTS")
print("=" * 45)

print("\n1. Total Passengers:", len(df))

print("\n2. Overall Survival Rate:",
      round(df['survived'].mean() * 100, 2), "%")

print("\n3. Gender-wise Survival Rate:")
print((df.groupby('sex')['survived'].mean() * 100).round(2))

print("\n4. Class-wise Survival Rate:")
print((df.groupby('pclass')['survived'].mean() * 100).round(2))

print("\n5. Average Passenger Age:",
      round(df['age'].mean(), 2))

print("\n6. Average Ticket Fare:",
      round(df['fare'].mean(), 2))

print("\n7. Most Common Passenger Class:",
      df['pclass'].mode()[0])

print("\n8. Highest Survival Rate by Gender:",
      df.groupby('sex')['survived'].mean().idxmax())

print("\n" + "=" * 45)
print("ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 45)
```

```python
df.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
